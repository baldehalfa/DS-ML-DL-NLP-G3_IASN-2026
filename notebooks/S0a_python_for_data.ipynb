## 3. Loops: doing something to every reading

You will never handle 52,000 readings one line at a time. A `for` loop does the same thing to each item in turn.

The **indentation is not decoration** - it is how Python knows which lines are inside the loop. Four spaces, consistently.

````python
for t in temps:
    print(f'reading: {t} C')

print('done - this line is outside the loop, so it runs once')
````

The name `t` is yours to choose; it takes each value in turn. Loops are most useful when they accumulate something:

````python
running_total = 0
for t in temps:
    running_total = running_total + t

print('total:', running_total)
````

## 7. Reading an error message

This is the most useful section in the notebook, and the one most courses skip.

You will cause errors constantly - everyone does, including people who write Python for a living. The difference between a fast session and a frustrating one is whether you *read* the error or panic at it.

**Two rules.**

1. **Read the last line first.** It names the problem.
2. **Then find the arrow.** It names the line that caused it.

Here are the four errors you will actually hit. Run each one and read it.

````python
print(gearbox_temperature)
````

**`NameError: name 'gearbox_temperature' is not defined`** - you used a name Python has never seen. Almost always a typo or a cell you did not run. The variable above is called `gearbox_temp`.

````python
print('temperature: ' + gearbox_temp)
````

**`TypeError: can only concatenate str (not "float") to str`** - you tried to add a number to text. Use an f-string instead: `print(f'temperature: {gearbox_temp}')`.

````python
print(temps[10])
````

**`IndexError: list index out of range`** - you asked for an item past the end. `len(temps)` is 6, so the valid positions are 0 to 5.

````python
print(machine['temperature'])
````

**`KeyError: 'temperature'`** - the dictionary has no such label. The key is `'gearbox_temp'`. Use `machine.keys()` when you are not sure.

Those four cover the large majority of what will go wrong for you in this course. **When something breaks, read the last line, then look at the arrow.**

## Summary

- `=` puts a value into a name. `==` asks whether two things are equal.
- Lists are ordered and **start at 0**; `[-1]` is the last item.
- `for` repeats; `if` decides; indentation says what is inside what.
- `def` names a calculation so you can reuse it; `return` hands back the answer.
- Dictionaries label their values, and you will mostly meet them as results.
- Read the **last line** of an error first, then find the arrow.

## Common mistakes

- **Off by one.** `temps[6]` on a six-item list. Valid positions are 0 to 5.
- **Missing quotes.** `turbine = T11` gives a `NameError`; text needs `'quotes'`.
- **Forgotten indentation.** The lines inside a loop or `if` must be indented, consistently.
- **`return` inside the loop.** It exits after the first item. Put it after the loop.
- **Adding text to a number.** Use an f-string: `f'temp is {t}'`.
